```python
import torch
```

```python
x = torch.tensor(1.0)
y = torch.tensor(2.0)

w = torch.tensor(1.0, requires_grad=True)
```

```python
print(x)
print(y)
```

```python
print(w)
```

```python
y_hat = w * x
s = y_hat - y
loss = (s) ** 2
```

```python
print(loss)
```

```python
print(loss.item())
```

```python
loss.backward()
```

```python
print(w.grad)
```

```python
print(w)
```

```python
print(w.grad)
```
